```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn as skl
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
# Load IRIS dataset
data = skl.datasets.load_iris()
```

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
```

```python
df.head()
```

```python
# Standardize the data
scaler = skl.preprocessing.StandardScaler()
```

```python
X = scaler.fit_transform(df)
```

```python
# Apply PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
```

```python
pca_X = pca.fit_transform(X)
```

```python
plt.scatter(pca_X[:, 0], pca_X[:, 1], c=data.target)
```

### Agglomerative Clustering


```python
# Construct Dendrogram

import scipy.cluster.hierarchy as sh

sh.dendrogram(sh.linkage(pca_X, method="ward"))
plt.title("Dendrogram")
plt.xlabel("Sample Index")
plt.ylabel("Euclidean Distance")
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=2, linkage="ward", metric="euclidean")
```

```python
cluster.fit(pca_X)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_X[:, 0], pca_X[:, 1], c=cluster.labels_)
```
